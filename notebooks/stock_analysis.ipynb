# Technical Indicators

```python
stocks = {
    "AAPL": df_aapl,
    "AMZN": df_amzn,
    "GOOG": df_goog,
    "META": df_meta,
    "NVDA": df_nvda
}

from ta.trend import SMAIndicator, EMAIndicator, MACD
from ta.momentum import RSIIndicator

for name, df in stocks.items():
    
    # Moving Averages
    df["SMA_20"] = SMAIndicator(close=df["Close"], window=20).sma_indicator()
    df["EMA_20"] = EMAIndicator(close=df["Close"], window=20).ema_indicator()
    
    # RSI
    df["RSI"] = RSIIndicator(close=df["Close"], window=14).rsi()
    
    # MACD
    macd = MACD(close=df["Close"])
    df["MACD"] = macd.macd()
    df["MACD_signal"] = macd.macd_signal()
    
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(3, 2, figsize=(16, 12))
axes = axes.flatten()

stock_list = ["AAPL", "AMZN", "GOOG", "META", "NVDA"]

for i, name in enumerate(stock_list):
    df = stocks[name]
    
    # filter to last 1 year for clarity
    df_filtered = df.iloc[-365:]
    
    axes[i].plot(df_filtered.index, df_filtered["Close"], label="Close", linewidth=1)
    axes[i].plot(df_filtered.index, df_filtered["SMA_20"], label="SMA 20")
    axes[i].plot(df_filtered.index, df_filtered["EMA_20"], label="EMA 20")
    
    axes[i].set_title(f"{name} - Moving Averages (Last 1 Year)")
    axes[i].legend()

axes[-1].axis("off")

plt.tight_layout()
plt.show()
```

```python
stock_list = ["AAPL", "AMZN", "GOOG", "META", "NVDA"]

for name in stock_list:
    df = stocks[name].dropna(subset=["SMA_20", "EMA_20"])
    
    above_sma = (df["Close"] > df["SMA_20"]).sum()
    below_sma = (df["Close"] <= df["SMA_20"]).sum()
    
    pct_above = above_sma / len(df) * 100
    
    last_close = df["Close"].iloc[-1]
    last_sma = df["SMA_20"].iloc[-1]
    
    current_pos = "ABOVE" if last_close > last_sma else "BELOW"
    gap_pct = abs((last_close - last_sma) / last_sma * 100)
    
    print("\n" + "=" * 60)
    print(f"{name} - MOVING AVERAGE INTERPRETATION")
    print("=" * 60)
    
    print(f"Days ABOVE SMA: {above_sma} ({pct_above:.1f}%)")
    print(f"Days BELOW SMA: {below_sma} ({100 - pct_above:.1f}%)")
    print()
    
    print(f"Last Close: {last_close:.2f}")
    print(f"SMA 20: {last_sma:.2f}")
    print(f"Current position: {current_pos} by {gap_pct:.2f}%")
    print()
    
    if pct_above > 60:
        print("Trend: Strong bullish bias (buyers dominant)")
    elif pct_above > 40:
        print("Trend: Sideways / indecisive market")
    else:
        print("Trend: Bearish bias (selling pressure dominant)")
    
    if current_pos == "ABOVE":
        print("Short-term signal: UPWARD momentum")
    else:
        print("Short-term signal: DOWNWARD momentum")
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(3, 2, figsize=(16, 12))
axes = axes.flatten()

stock_list = ["AAPL", "AMZN", "GOOG", "META", "NVDA"]

for i, name in enumerate(stock_list):
    df = stocks[name]
    
    axes[i].plot(df.index, df["RSI"], label="RSI")
    
    axes[i].axhline(70, linestyle="--", color="red", linewidth=1)
    axes[i].axhline(30, linestyle="--", color="green", linewidth=1)
    
    axes[i].set_title(f"{name} - RSI")
    axes[i].legend()

# turn off empty subplot (6th slot)
axes[-1].axis("off")

plt.tight_layout()
plt.show()
```

```python
stock_list = ["AAPL", "AMZN", "GOOG", "META", "NVDA"]

for name in stock_list:
    df = stocks[name].dropna(subset=["RSI"])
    
    overbought = (df["RSI"] > 70).sum()
    oversold = (df["RSI"] < 30).sum()
    neutral = ((df["RSI"] >= 30) & (df["RSI"] <= 70)).sum()
    
    total = len(df)
    
    print("\n" + "=" * 60)
    print(f"{name} - RSI INTERPRETATION")
    print("=" * 60)
    
    print(f"Overbought days (RSI > 70): {overbought} ({overbought/total*100:.1f}%)")
    print(f"Oversold days (RSI < 30): {oversold} ({oversold/total*100:.1f}%)")
    print(f"Neutral days: {neutral} ({neutral/total*100:.1f}%)")
    print()
    
    last_rsi = df["RSI"].iloc[-1]
    print(f"Current RSI: {last_rsi:.2f}")
    
    if last_rsi > 70:
        print("Signal: Overbought -> possible pullback or correction")
    elif last_rsi < 30:
        print("Signal: Oversold -> possible rebound or recovery")
    else:
        print("Signal: Neutral -> no extreme momentum condition")
```

```python
stock_list = ["AAPL", "AMZN", "GOOG", "META", "NVDA"]

for name in stock_list:
    df = stocks[name].dropna(subset=["MACD", "MACD_signal"])
    
    # when MACD is above signal line
    bullish_periods = (df["MACD"] > df["MACD_signal"]).sum()
    
    # when MACD is below signal line
    bearish_periods = (df["MACD"] < df["MACD_signal"]).sum()
    
    total = len(df)
    
    print("\n" + "=" * 60)
    print(f"{name} - MACD INTERPRETATION")
    print("=" * 60)
    
    print(f"Bullish periods (MACD > Signal): {bullish_periods} ({bullish_periods/total*100:.1f}%)")
    print(f"Bearish periods (MACD < Signal): {bearish_periods} ({bearish_periods/total*100:.1f}%)")
    print()
    
    last_macd = df["MACD"].iloc[-1]
    last_signal = df["MACD_signal"].iloc[-1]
    
    print(f"Current MACD: {last_macd:.4f}")
    print(f"Current Signal: {last_signal:.4f}")
    
    if last_macd > last_signal:
        print("Signal: Bullish momentum -> upward trend likely continuing")
    else:
        print("Signal: Bearish momentum -> downward pressure or weakening trend")
```

# Additional financial metrics

Pynance installation was giving problems so TA library is used instead.

```python
stock_list = ["AAPL", "AMZN", "GOOG", "META", "NVDA"]

for name in stock_list:
    df = stocks[name]
    
    df["Daily_Return"] = df["Close"].pct_change()
    
```

```python
import matplotlib.pyplot as plt

stock_list = ["AAPL", "AMZN", "GOOG", "META", "NVDA"]

fig, axes = plt.subplots(3, 2, figsize=(16, 12))
axes = axes.flatten()

for i, name in enumerate(stock_list):
    df = stocks[name]
    
    axes[i].plot(df.index, df["Daily_Return"])
    
    axes[i].set_title(f"{name} - Daily Returns")
    axes[i].set_xlabel("Date")
    axes[i].set_ylabel("Return")

# remove empty subplot
axes[-1].axis("off")

plt.tight_layout()
plt.show()
```

```python
stock_list = ["AAPL", "AMZN", "GOOG", "META", "NVDA"]

for name in stock_list:
    df = stocks[name].dropna(subset=["Daily_Return"])
    
    avg_return = df["Daily_Return"].mean()
    std_return = df["Daily_Return"].std()
    
    max_gain = df["Daily_Return"].max()
    max_loss = df["Daily_Return"].min()
    
    print("\n" + "=" * 60)
    print(f"{name} - DAILY RETURNS INTERPRETATION")
    print("=" * 60)
    
    print(f"Average daily return: {avg_return:.5f}")
    print(f"Volatility (std dev): {std_return:.5f}")
    print()
    
    print(f"Best single-day gain: {max_gain:.2%}")
    print(f"Worst single-day loss: {max_loss:.2%}")
    print()
    
    if std_return > 0.03:
        print("Risk level: HIGH volatility (large price swings)")
    elif std_return > 0.015:
        print("Risk level: MODERATE volatility")
    else:
        print("Risk level: LOW volatility (stable asset)")
```

```python
import matplotlib.pyplot as plt

stock_list = ["AAPL", "AMZN", "GOOG", "META", "NVDA"]

fig, axes = plt.subplots(3, 2, figsize=(16, 12))
axes = axes.flatten()

for i, name in enumerate(stock_list):
    df = stocks[name]
    
    df["Volatility_20"] = (
        df["Daily_Return"]
        .rolling(window=20)
        .std()
    )
    
    axes[i].plot(df.index, df["Volatility_20"])
    
    axes[i].set_title(f"{name} - 20 Day Volatility")
    axes[i].set_xlabel("Date")
    axes[i].set_ylabel("Volatility")

# remove empty subplot
axes[-1].axis("off")

plt.tight_layout()
plt.show()
```

```python
stock_list = ["AAPL", "AMZN", "GOOG", "META", "NVDA"]

for name in stock_list:
    df = stocks[name].dropna(subset=["Volatility_20"])
    
    avg_vol = df["Volatility_20"].mean()
    max_vol = df["Volatility_20"].max()
    min_vol = df["Volatility_20"].min()
    
    print("\n" + "=" * 60)
    print(f"{name} - VOLATILITY INTERPRETATION")
    print("=" * 60)
    
    print(f"Average volatility: {avg_vol:.5f}")
    print(f"Peak volatility: {max_vol:.5f}")
    print(f"Lowest volatility: {min_vol:.5f}")
    print()
    
    if avg_vol > 0.03:
        print("Risk level: HIGH (very unstable price behavior)")
    elif avg_vol > 0.015:
        print("Risk level: MODERATE (some fluctuations)")
    else:
        print("Risk level: LOW (relatively stable price movement)")
    
    print("Interpretation: Volatility measures uncertainty and risk in price movements over time.")
```
